```python
#3 Flight Scheduling and Pricing application [Gene Expression Algorithm (GEA)]
import random

CAPACITY, BASE_FARE, FIXED_COST = 180, 150.0, 12000.0
PEAK_HOURS = {7, 8, 9, 17, 18, 19}

def evaluate(chrom):
    hour, p_mult = chrom[0] % 24, 0.8 + 1.2 * ((chrom[1] % 100) / 100.0)
    base_demand = 150 * (1.6 if hour in PEAK_HOURS else 0.8)
    passengers = min(CAPACITY, base_demand * (p_mult ** -1.5))
    revenue = passengers * (BASE_FARE * p_mult)
    cost = FIXED_COST + (2500.0 if hour in PEAK_HOURS else 1000.0)
    return max(0.0, revenue - cost), hour, p_mult, passengers

pop = [[random.randint(0, 200) for _ in range(4)] for _ in range(50)]

for gen in range(40):
    evals = [evaluate(ind) for ind in pop]
    scores = [e[0] for e in evals]
    
    parents = [pop[i] if scores[i] > scores[j] else pop[j] 
               for i, j in [random.sample(range(50), 2) for _ in range(50)]]
    
    next_pop = []
    for i in range(0, 50, 2):
        c1, c2 = parents[i][:], parents[i+1][:]
        if random.random() < 0.8: # Crossover
            c1[2:], c2[2:] = parents[i+1][2:], parents[i][2:]
        for c in (c1, c2): # Mutate
            if random.random() < 0.1: c[random.randint(0, 3)] = random.randint(0, 200)
        next_pop.extend([c1, c2])
    pop = next_pop

best_profit, hour, price_m, pax = max([evaluate(ind) for ind in pop], key=lambda x: x[0])
print(f"Optimal Departure: {hour:02d}:00 | Ticket Price: ${BASE_FARE * price_m:.2f}")
print(f"Passengers: {pax:.0f}/{CAPACITY} | Max Profit: ${best_profit:,.2f}")
```
