```python
#1 using password guesser application [Genetic Algorithm]
import random, string

TARGET = "HELLO GAS"
CHARS = string.ascii_uppercase + " "

def fitness(cand): 
    return sum(1 for c, t in zip(cand, TARGET) if c == t)

pop = [''.join(random.choices(CHARS, k=len(TARGET))) for _ in range(100)]

for gen in range(100):
    pop.sort(key=fitness, reverse=True)
    if fitness(pop[0]) == len(TARGET):
        print(f" Solved in Generation {gen}: '{pop[0]}'")
        break
         
    survivors = pop[:20]
    next_pop = survivors[:]
    while len(next_pop) < 100:
        p1, p2 = random.sample(survivors, 2)
       
        child = p1[:4] + p2[4:]
        
        child = ''.join(c if random.random() > 0.05 else random.choice(CHARS) for c in child)
        next_pop.append(child)
    pop = next_pop
```
