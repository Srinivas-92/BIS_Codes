```python
#2 network srenght check application [Partical Swarm Optimization (PSO)]
import random


def fitness(x, y):
    return 100 - ((x - 70)**2 + (y - 80)**2)**0.5

particles = [{'pos': [random.uniform(0, 100), random.uniform(0, 100)], 
              'vel': [0, 0], 'pbest_pos': [], 'pbest_val': float('-inf')} for _ in range(5)]

gbest_pos, gbest_val = None, float('-inf')
for p in particles:
    p['pbest_pos'] = list(p['pos'])
    p['pbest_val'] = fitness(p['pos'][0], p['pos'][1])
    if p['pbest_val'] > gbest_val:
        gbest_val, gbest_pos = p['pbest_val'], list(p['pos'])

for it in range(5):
   
    for p in particles:
        for i in range(2):
            p['vel'][i] = 0.5 * p['vel'][i] + 1.0 * random.random() * (p['pbest_pos'][i] - p['pos'][i]) + 1.0 * random.random() * (gbest_pos[i] - p['pos'][i])
            p['pos'][i] += p['vel'][i]

    for p in particles:
        val = fitness(p['pos'][0], p['pos'][1])
        if val > p['pbest_val']:
            p['pbest_val'], p['pbest_pos'] = val, list(p['pos'])
        if val > gbest_val:
            gbest_val, gbest_pos = val, list(p['pos'])

    print(f"Iteration {it+1}: Best Signal={gbest_val:.1f} at Pos={[round(c, 1) for c in gbest_pos]}")

print(f"\nFinal Result: Best Signal Found = {gbest_val:.1f} at Location {[round(c, 1) for c in gbest_pos]}")
```
